```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
```

# StandardScallar
formula:
X(i) = X(i) - mean / standard-deviation

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
scaler.fit(X_train)
scalled_X_train = scaler.transform(X_train)
scalled_X_test = scaler.transform(X_test)
```

```python
scalled_X_test
```

```python
scalled_X_train
```

```python
# Mean
scaler.mean_
```

```python
# std deviation
scaler.scale_
```

```python
# variance
scaler.var_
```

# Convert Numpy array into Dataframe

```python
scalled_X_train = pd.DataFrame(scalled_X_train, columns=X_train.columns)
scalled_X_test = pd.DataFrame(scalled_X_test, columns=X_test.columns)
```

```python
scalled_X_train
```

```python
scalled_X_test
```

```python
np.round(X_train.describe(),1)
```

When we apply standardisation we convert our mean to --> 'Zero"
and std deviation into --. "1"

```python
np.round(scalled_X_train.describe(),1)
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr_scalled = LinearRegression()
```

```python
lr.fit(X_train, y_train)
lr_scalled.fit(scalled_X_train, y_train)
```

```python
lr_pred = lr.predict(X_test)
lr_scalled_pred = lr_scalled.predict(scalled_X_test)
```

```python
from sklearn.metrics import r2_score
```

```python
print(r2_score(y_test, lr_pred))
print(r2_score(y_test, lr_scalled_pred))
```
