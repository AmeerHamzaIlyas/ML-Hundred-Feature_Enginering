# 2. By using sciket learn

```python
df
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(df.iloc[:, 0:2], df.iloc[:, -1], test_size=0.2, random_state=2)
```

```python
X_train
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe = OneHotEncoder(drop='first', sparse_output=False, dtype=np.int32)
# drop='first' --> to avoid multicollinearity problem
# sparse_output=False --> to return a dense numpy array instead of a sparse matrix
# dtype=np.int32 --> to convert float to int32
```

```python
ohe.fit(X_train[["fuel", "brand"]])
```

```python
ohe_X_train = ohe.transform(X_train[["fuel", "brand"]])
ohe_X_test = ohe.transform(X_test[["fuel", "brand"]])
```

```python
ohe_X_train
```

```python
ohe_X_test
```

```python
X_train.shape
```

```python
X_train_new = np.hstack((y_train.values.reshape(-1, 1), ohe_X_train))
X_test_new = np.hstack((y_test.values.reshape(-1, 1), ohe_X_test))
```
